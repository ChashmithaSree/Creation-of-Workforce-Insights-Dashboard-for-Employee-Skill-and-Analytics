Create a class called Book :

The __init__ method should accept title, author, and price, and store them as attributes.
Add a method apply_discount(percent) that reduces self.price by the given percentage.
Add a __str__ method so that printing a Book object shows something like: "Atomic Habits" by James Clear - ₹500

```python
class Book:
  def __init__(self,title,author,price):
    self.title = title
    self.author = author
    self.price = price
  def apply_discount(self,percent):
    self.price = self.price - self.price * (percent/100)
  def __str__(self):
    return f'"{self.title}" by {self.author} - ${self.price}'
b = Book("Atomic Habits", "James", 500)
print(b)
b.apply_discount(20)
print(b)
```

Create a class called Rectangle :

The __init__ method should accept length and width, and store them as attributes.
Add a method area() that returns length × width.
Add a method perimeter() that returns 2 × (length + width).
Add a class method square(cls, side) that creates and returns a Rectangle where length = width = side

```python
class Rectangle:
  def __init__(self,length,width):
    self.length = length
    self.width = width
  def perimeter(self):
    return 2 * (self.length+self.width)
  def area(self):
    return self.length * self.width
  def __str__(self):
    return f'Area:{self.area()}, Perimeter:{self.perimeter()}'
a = Rectangle(10,20)
print(a)
```

Create a class called BankAccount :

The __init__ method should accept owner and balance (default 0), and store them as attributes.
Add a method deposit(amount) that adds amount to the balance.
Add a method withdraw(amount) that subtracts amount from the balance, but should not allow withdrawing more than the current balance (raise an error or return a message like "Insufficient Balance").
Add a __str__ method so printing the account shows something like: tam's account: ₹1000

```python
class BankAccount:
  def __init__(self,owner,balance):
    self.owner = owner
    self.balance = balance
  def deposit(self,amount):
   self.balance = self.balance+amount
  def withdraw(self,amount):
    if self.balance > amount:
      self.balance = self.balance-amount
    else:
      return f'Insufficient Balance'
  def __str__(self):
    return f'{self.owner} account:${self.balance}'
a = BankAccount("tam",2500)
print(a)
a.deposit(500)
print(a)
a.withdraw(500)
print(a)
```

Create a class called Temperature:

The __init__ method should accept celsius and store it as an attribute.
Add a fahrenheit property (using @property) that returns the Celsius value converted to Fahrenheit: F = C * 9/5 + 32
Add a setter for fahrenheit (using @fahrenheit.setter) so that setting fahrenheit automatically updates self.celsius using: C = (F - 32) * 5/9



```python
class Temperature:
  def __init__(self,celsius):
    self.celsius = celsius
  @property
  def farenheit(self):
   return self.celsius *(9/5) + 32
  @farenheit.setter
  def farenheit(self,value):
    self.celsius = (value-32) * 5/9
  def __str__(self):
    return f'Celsius:{self.celsius}C Farenheit:{self.farenheit}F'
a = Temperature(30)
print(a.celsius)
print(a.farenheit)
print(a)
```
