Write pure SQL to create a table called "students" with columns: emp_id (INTEGER), name (TEXT), department (TEXT). Insert 4 sample rows with realistic student names and departments (e.g., aiml, aids, cs, cse). Then write a SELECT query to display all the records in the students table.

Only output the SQL code, no explanation.

```python
%%sql
DROP TABLE IF EXISTS students;

CREATE TABLE students(
    id INTEGER,
    name TEXT,
    dept TEXT
);
```
