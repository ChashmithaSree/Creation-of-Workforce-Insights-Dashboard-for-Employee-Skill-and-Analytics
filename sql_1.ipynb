Write a sql code to join the learners table and departments table.
Write a SELECT query that shows learners name,skill_score and department head's name ,by matching learners.department to the departments.dept_name.

```python
%%sql
SELECT learners.name, learners.skill_score, departments.head
FROM learners
JOIN departments ON learners.department = departments.dept_name;
```

Write a sql code to show avg skillscore  of departments greater than 7

```python
%%sql
SELECT department, AVG(skill_score) as avg_score FROM learners GROUP BY department HAVING AVG(skill_score) > 7;
```

Write a select query to show names that start with "ch" from learners table


```python
%%sql
SELECT * FROM learners WHERE name LIKE 'ch%';
```

write a select query to show students whose skill score is in between 6 and 9 from the learners table

```python
%%sql
SELECT * FROM learners WHERE skill_score BETWEEN 6 AND 9;
```
